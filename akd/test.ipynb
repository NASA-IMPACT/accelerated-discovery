```python
class MyClass:
    class_var = "I'm a class variable"

    def __init__(self, instance_var):
        self.instance_var = instance_var

    def regular_method(self):
        # This method operates on an instance
        print(f"I'm a regular method. Instance variable: {self.instance_var}")
        print(f"I can also access the class variable: {MyClass.class_var}")

    @classmethod
    def class_method(cls):
        # This method operates on the class itself
        print(f"I'm a class method. The first argument is the class: {cls}")
        print(f"I can access the class variable: {cls.class_var}")
        # Note: You cannot access instance_var here without an instance
```

```python
        
# Creating an instance
obj = MyClass("I'm an instance variable")

# Calling the methods
obj.regular_method() 
```
